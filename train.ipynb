```python
import jieba
import os
from gensim.models import Word2Vec

def load_corpus(corpus_path=None):
    """加载语料，内置验证语料避免文件问题"""
    if corpus_path and os.path.exists(corpus_path):
        with open(corpus_path, "r", encoding="utf-8") as f:
            lines = [line.strip() for line in f if line.strip()]
    else:
        # 内置验证语料（确保包含"数据"等词）
        lines = [
            "数据科学是利用科学方法从数据中提取知识的领域",
            "Python 是一种广泛使用的高级编程语言，适合数据分析与人工智能开发",
            "人工智能是计算机科学的一个分支，旨在模拟人类智能",
            "机器学习是人工智能的核心技术，通过数据训练模型实现预测",
            "代码是实现算法和构建项目的基础工具",
            "数据可视化能将抽象数据转化为直观图表",
            "项目实践需要用到数据和模型",
            "环境配置是数据项目的第一步"
        ]
    corpus = [list(jieba.cut(line)) for line in lines]
    return corpus

def train_word2vec(corpus_path=None, model_path="models/word2vec.model"):
    """训练Word2Vec模型并保存"""
    os.makedirs(os.path.dirname(model_path), exist_ok=True)
    corpus = load_corpus(corpus_path)
    
    # 训练模型
    model = Word2Vec(
        sentences=corpus,
        vector_size=100,
        window=5,
        min_count=1,
        workers=4,
        epochs=10
    )
    model.save(model_path)
    print(f"✅ 模型已保存至: {model_path}")
    return model
```
