```python
import numpy as np
import matplotlib.pyplot as plt
from gensim.models import Word2Vec
from sklearn.decomposition import PCA

def visualize_words(model_path, words):
    # 1. 中文显示配置（解决文字乱码）
    plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei']
    plt.rcParams['axes.unicode_minus'] = False

    # 2. 加载模型和词向量
    model = Word2Vec.load(model_path)
    word_vectors = np.array([model.wv[word] for word in words])
    pca = PCA(n_components=2)
    vec_2d = pca.fit_transform(word_vectors)

    # 3. 创建画布并修复坐标轴方向
    plt.figure(figsize=(10, 8))
    ax = plt.gca()
    # 反转异常的坐标轴
    ax.invert_xaxis()
    ax.invert_yaxis()

    # 4. 绘制点和文字（文字紧贴点，方向正常）
    for i, word in enumerate(words):
        x, y = vec_2d[i]
        plt.scatter(x, y, s=150, alpha=0.7)
        # 极小偏移，文字紧贴点的右上方，方向正常
        plt.text(x + 0.002, y + 0.002, word, fontsize=11, ha='left', va='bottom')

    # 5. 调整布局和标题
    plt.title("Word2Vec 词向量分布可视化（PCA降维）", fontsize=14, pad=15)
    plt.tight_layout()
    plt.savefig("word2vec_visualization.png", dpi=300, bbox_inches='tight')
    plt.show()
```
